```python
x = torch.tensor(2.0, requires_grad=True)

y = x**2 + 3 * x
y.backward()

print(x.grad)  # dy/dx = 2x + 3 = 7
```

```python
import torch

x = torch.tensor(2.0)
w = torch.tensor(3.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)
y = torch.tensor(10.0)

z = w * x + b
loss = (z - y) ** 2

loss.backward()

print(w.grad)
print(b.grad)
```

```python
w = torch.tensor(0.5, requires_grad=True)
x= torch.tensor(1)
b= torch.tensor(0.5,requires_grad=True)
t= torch.tensor(1)


c = w*x
d = c+b
yhat= torch.sigmoid(d)
loss = (yhat-t)**2

loss.backward()

print(w.grad)
print(b.grad)


```
